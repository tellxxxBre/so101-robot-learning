# SO101 grape色分け 本番学習（Colab / GPU用）

**事前準備（Colabを開く前に）**
1. Macで作った `so101_grape_sort_dataset.zip`（`outputs/`フォルダの中にあります）を、Googleドライブの `MyDrive/so101/` フォルダにアップロードしてください
2. Colab上部メニュー「ランタイム」→「ランタイムのタイプを変更」→ハードウェアアクセラレータを **GPU** にする（有料のコンピューティングユニットがあれば A100 か V100、無ければ T4）
3. 上から順にセルを実行（Shift+Enter）してください

**注意**：エピソード49（赤ブドウの受け渡しが外れて失敗）は`--dataset.episodes`で除外し、0〜48の49エピソードだけを学習に使います。

```
# GPUがちゃんと割り当てられているか確認
!nvidia-smi
```

```
from google.colab import drive
drive.mount('/content/drive')
```

```
# アップロードしたzipを解凍（パスが違う場合はここを修正してください）
DRIVE_ZIP = "/content/drive/MyDrive/so101/so101_grape_sort_dataset.zip"

!mkdir -p /content/dataset
!unzip -q "$DRIVE_ZIP" -d /content/dataset
!ls /content/dataset
```

```
# lerobotをMacと同じバージョンでインストール（[dataset]で映像処理用のavも一緒に入れる）
!pip install -q "lerobot[dataset,training]==0.6.0" accelerate
```

```
# 解凍したデータセットの実際のフォルダ名を自動検出
import glob

candidates = glob.glob("/content/dataset/so101_grape_sort_*")
assert len(candidates) == 1, f"データセットフォルダが1つに絞れません: {candidates}"
DATASET_ROOT = candidates[0]
print("DATASET_ROOT =", DATASET_ROOT)
```

```
# 本番学習（100,000ステップ）。エピソード49（受け渡し失敗）は除外して0〜48の49エピソードのみ使用。
# output_dirを最初からGoogleドライブにして、5000ステップごとのチェックポイントが都度残るようにしている。
EPISODES = "[" + ",".join(str(i) for i in range(49)) + "]"

!lerobot-train \
  --dataset.repo_id=local/so101_grape_sort \
  --dataset.root="$DATASET_ROOT" \
  --dataset.episodes="$EPISODES" \
  --policy.type=act \
  --policy.device=cuda \
  --policy.push_to_hub=false \
  --output_dir=/content/drive/MyDrive/so101/train_output/act_grape_sort \
  --job_name=act_grape_sort \
  --steps=100000 \
  --batch_size=8 \
  --save_freq=5000 \
  --log_freq=50 \
  --wandb.enable=false
```

```
# output_dirが最初からGoogleドライブなので、ここでのコピー作業は不要。
# 完了確認だけ行う（チェックポイント一覧が表示されればOK）
!ls -la /content/drive/MyDrive/so101/train_output/act_grape_sort/checkpoints
```
